```python
import xarray
import numpy as np
import pickle
import sys
sys.path.append('../get_planet_params/')
import distributions
```

```python
with open('TOI455_wl_T2_nrs1.pkl', 'rb') as file:
    res_nrs1 = pickle.load(file)
with open('TOI455_wl_T2_nrs2.pkl', 'rb') as file:
    res_nrs2 = pickle.load(file)
```

```python
bins = np.loadtxt('DefaultBins.txt')
short = bins[:, 0]
long = bins[:, 1]

spect = np.hstack([res_nrs1[0]['spect'], res_nrs2[0]['spect']])
wavs = np.hstack([res_nrs1[0]['wavs'], res_nrs2[0]['wavs']])

spectro_lcs = []
for s, l in zip(short, long):
    mask = (wavs > s) & (wavs < l)
    #print(np.sum(spect[:, mask], axis=1).shape)
    spectro_lcs.append(np.sum(spect[:, mask], axis=1))
    
spectro_lcs = np.array(spectro_lcs)
```

```python
np.savetxt('spectroscopic_lcs.txt', spectro_lcs)
```
